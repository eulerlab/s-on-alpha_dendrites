# Imports

```python
import os

import warnings
from djimaging.user.alpha.utils import access_database

# Choose indicator by uncommenting one of the following lines
indicator = 'calcium'
# indicator = 'glutamate'

database = access_database.access_database(indicator, use_sql=True)
```

# Data Base

```python
import pandas as pd

output_dir = "/gpfs01/berens/user/joesterle/projects/alpha/__public_s-on-alpha-regional/gnode/s-on-alpha-regional/database-exports"

assert os.path.isdir(output_dir)

def export_table(tab, filename='output_file', output_dir=output_dir):
    if not isinstance(tab, pd.DataFrame):
        df = tab.fetch(format='frame')
    else:
        df = tab
    df = df.infer_objects()

    if not filename.endswith('.h5'):
        filename = filename + '.h5'

    filepath = os.path.join(output_dir, filename)

    with warnings.catch_warnings():
        warnings.simplefilter('ignore')
        df.to_hdf(filepath, key='df', mode='w')

    return filepath
```
